## Imports

```python
import pandas as pd
import numpy as np
```

## Data

```python
train = pd.read_parquet("data/train.parquet", engine="fastparquet")
target = pd.read_parquet("data/target_user_ids.parquet", engine="fastparquet")
```

```python
train.sample(5)
```

```python
target.sample(5)
```

## EDA train dataset 

### Top level

```python
train["date"].describe()
```

```python
train["event_type"].value_counts()
```

### By events

```python
favorite = train[train["event_type"] == "favorite"]
favorite.describe()
```

```python
like = train[train["event_type"] == "like"]
like.describe()
```

```python
watch_time = train[train["event_type"] == "watch_time"]
watch_time.describe()
```

```python
watch_time.watch_time.hist()
```

```python
np.log(watch_time.watch_time).hist()
```

```python
watch_time.loc[:, "watch_time"] = np.log(watch_time.watch_time + 1)
```

## EDA train_user_ids

```python
count_events = train.groupby(["user_id"])["item_id"].count()
count_events
```

```python
merged_target = target.merge(count_events, how="left", on="user_id")
```

```python
merged_target.item_id.hist()
```

```python
merged_target.describe()
```
